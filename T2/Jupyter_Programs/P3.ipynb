# WAP to encrypt & decrypt your message using the shift key

```python
s=input("Enter String:")
key=int(input("key="))
s1=""

#Encryption
for i in range(key):
    s1+=s[i::key]
print(s1)

#Decryption
length=len(s1)
part=length//key
extra=length%key
part1=s1[:(part+1)*extra]
part2=s1[(part+1)*extra:]
msg=""
for i in range(part+1):
    if i<part:
        msg+=part1[i::part+1]+part2[i::part]
    else:
        msg+=part1[i::part+1]
print("Decrypted msg:",msg)
```

# aliasing & cloning

```python
x=[1,2,3]
y=x
print(id(x))
print(id(y))
y[1]=99
print(x)
print(y)
```

```python
# clonnig 
# method-1
x=[1,2,3]
y=x[:]
y[0]=99
print(x)
print(y)

# method-2
z=x.copy()
z[1]=100
print(z)
print(x)
```
