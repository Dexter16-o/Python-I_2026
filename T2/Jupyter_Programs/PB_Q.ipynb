```python
L1 = [1, 1.33, 'LJU', 0, 'N', True, 'Y', 1]
val1= 0
val2= ""
for x in L1:
    if(type(x) == int or type(x) == float):
        val1 += x
    elif(type(x) == str):
        val2 += x
    else:
        break
        continue
print(val1, val2)
```
