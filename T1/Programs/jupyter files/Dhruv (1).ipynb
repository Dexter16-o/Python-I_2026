```python
#2.Relational Operators: >,<,>=,<=,==,!=
print(10>0)
print(10<0)
print(True<False)
print(True==False)
print(True>False)
print(10>=10)
print(10<=5)
print(10!=10)
print(10==10.0)
print(10<20<30)
print(10>20<30)
print('ab'=='ab')
print('AB'=='ab')
print(10>True)
print(0>False)
```

```python
#3.Logical Operators: and,or,not
#and: if both true -> true
#or: if both false -> false
#not: invert

print(False and True)
print(False and False)
print(True and True)
print(True and False)
print(True or True)
print(False or True)
print(True or False)
print(False or False)
print("1",10 and 20)
print("2",0 and 20)
print("3",20 and 0)
print("4",10 or 20)
print("5",0 or 10)
print("6",10 or 0)
print("7",not 20)
print("8",not 0)
print("9",'ab' or 'cd')
print("10",'' and 'cd')
print("11",'ab' or ' ')
```

```python
#4.Conditional(Ternary) operators
a,b=10,20
x=30 if a>b else 40 
print(x)
```
